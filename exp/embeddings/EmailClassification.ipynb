```python
from sentence_transformers import SentenceTransformer
from huggingface_hub import notebook_login

import pandas as pd

notebook_login()
```

```python
# Download from the 🤗 Hub
ads = pd.read_csv("emails.csv")

model = SentenceTransformer("google/embeddinggemma-300m")
model.
#model = SentenceTransformer("Qwen/Qwen3-Embedding-4B")
```

```python
query_embeddings = model.encode_query("Rejection", prompt_name="Classification")

print(query_embeddings)
ads["Results"] = ads["Text"].apply(lambda x: model.similarity(query_embeddings, model.encode_document(x, prompt_name="Classification")).item())

print(ads.sort_values("Results", ascending=False)[["Email", "Results"]])
```

```python
query_embeddings = model.encode_query("Invitation letter", prompt_name="Classification")

ads["Results"] = ads["Text"].apply(lambda x: model.similarity(query_embeddings, model.encode_document(x, prompt_name="Classification")).item())

print(ads.sort_values("Results", ascending=False))
```

```python
query_embeddings = model.encode_query("Is it letter from EchoHub, about Backend Engineer position?", prompt_name="query")

ads["Results"] = ads["Text"].apply(lambda x: model.similarity(query_embeddings, model.encode_document(x)).item())

print(ads.sort_values("Results", ascending=False))
```
